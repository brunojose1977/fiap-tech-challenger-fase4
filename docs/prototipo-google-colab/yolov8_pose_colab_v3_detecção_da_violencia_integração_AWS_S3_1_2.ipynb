## Modelo

Variantes comuns: `yolov8n-pose.pt` (rápido), `yolov8s-pose.pt`, `yolov8m-pose.pt`, `yolov8l-pose.pt`, `yolov8x-pose.pt` (mais preciso, mais lento). O primeiro download é automático.

```python
MODEL_NAME = "yolov8n-pose.pt"  # troque por s/m/l/x se tiver GPU e quiser mais qualidade
model = YOLO(MODEL_NAME)
```

## Inferência no vídeo

- `conf`: limiar de confiança das detecções.
- `imgsz`: tamanho de entrada (múltiplo de 32 costuma funcionar bem).
- `device=0` força GPU; em CPU omita ou use `device="cpu"`.

```python
from pathlib import Path

import boto3

results = model.predict(
    source=str(VIDEO_PATH),
    task="pose",
    save=True,
    conf=0.25,
    iou=0.7,
    imgsz=640,
    stream=True,
    verbose=True,
)

out_path = None
for r in results:
    out_path = getattr(r, "save_dir", None)

print("Pasta de saída:", out_path)

# Upload do vídeo gerado pelo predict para o bucket S3 yolo-predict-tcf4
predict_bucket = "yolo-predict-tcf4"
if out_path:
    save_dir = Path(out_path)
    predict_video = None
    for pattern in ("*.avi", "*.mp4"):
        found = list(save_dir.glob(pattern))
        if found:
            predict_video = max(found, key=lambda p: p.stat().st_mtime)
            break
    if predict_video is not None and predict_video.is_file():
        if not all([aws_access_key_id, aws_secret_access_key, aws_default_region]):
            print("S3: credenciais AWS ausentes; vídeo de predict não foi enviado.")
        else:
            s3_client = boto3.client(
                "s3",
                aws_access_key_id=aws_access_key_id,
                aws_secret_access_key=aws_secret_access_key,
                region_name=aws_default_region,
            )
            s3_key = predict_video.name
            print(f"Enviando '{predict_video}' para s3://{predict_bucket}/{s3_key} ...")
            s3_client.upload_file(str(predict_video), predict_bucket, s3_key)
            print(f"Upload concluído: s3://{predict_bucket}/{s3_key}")
    else:
        print(f"S3: nenhum .avi/.mp4 encontrado em {save_dir}")
else:
    print("S3: save_dir desconhecido; vídeo de predict não enviado.")
```

### Violence Detection Logic (Conceptual)

Detecting "violence" from pose estimation data is a complex task that typically involves analyzing keypoint movements, velocities, joint angles, and spatial relationships over time. The `ultralytics` model provides raw keypoint data, but the interpretation for specific actions like "violence" needs custom logic or a specialized model trained on such behaviors.

Below is a **placeholder function** for `is_violence_detected`. This example uses a very simple (and likely inadequate for real-world use) condition for demonstration. **You will need to replace this with more robust logic** based on your definition of violence (e.g., specific limb configurations, rapid changes in keypoint positions, or relative positions between multiple individuals).

```python
import numpy as np

def is_violence_detected(person_keypoints, confidence_threshold=0.5):
    """
    Placeholder function to detect violence based on keypoints.
    This is a simplified example; real violence detection requires more sophisticated logic.

    Args:
        person_keypoints (ultralytics.engine.results.Keypoints): Keypoints object for a single person.
        confidence_threshold (float): Minimum confidence for a keypoint to be considered.

    Returns:
        bool: True if violence is 'detected' (based on placeholder logic), False otherwise.
    """
    if person_keypoints is None or person_keypoints.data.numel() == 0:
        return False

    # Example placeholder logic: If a person has a high number of visible keypoints
    # with confidence above a certain threshold, we'll 'trigger' detection.
    # This is NOT actual violence detection, but a demo of where to put your logic.
    kpts = person_keypoints.data[0] # (17, 3) -> x, y, conf for 17 keypoints
    visible_kpts = kpts[kpts[:, 2] > confidence_threshold] # Filter by confidence

    # If more than 10 keypoints are visible with high confidence, consider it a 'trigger'
    # Replace this with actual violence detection logic (e.g., analyze angles, velocities, etc.)
    if len(visible_kpts) > 10:
        return True

    return False

print("Violence detection placeholder function defined.")
```

### Re-running Inference with Violence Detection

Now, let's re-run the pose estimation inference and apply our conceptual violence detection logic to each detected person's keypoints. If violence is 'detected' in any frame, we will print a red message.

```python
import cv2
import boto3
from pathlib import Path

print("\033[96mStarting inference with violence detection and video annotation...\033[0m") # Cyan color

# Get video properties for VideoWriter
cap = cv2.VideoCapture(str(VIDEO_PATH))
if not cap.isOpened():
    raise IOError(f"Error: Could not open video file {VIDEO_PATH}")

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
cap.release() # Release the capture object as we don't need it for frame reading here

# Define output video path
output_video_dir = Path("/content/runs/pose/custom_annotated_videos")
output_video_dir.mkdir(parents=True, exist_ok=True)
output_video_path = output_video_dir / f"video_violence_detected_{MODEL_NAME.replace('.pt', '')}.mp4"

# Initialize VideoWriter
# Choose an appropriate codec, e.g., 'mp4v' for .mp4, 'XVID' for .avi
fourcc = cv2.VideoWriter_fourcc(*'mp4v') # Codec for .mp4 files
out_video = cv2.VideoWriter(str(output_video_path), fourcc, fps, (width, height))

if not out_video.isOpened():
    raise IOError(f"Error: Could not create video writer for {output_video_path}")
else:
    print(f"Output video will be saved to: {output_video_path}")


# Re-run the prediction to get a fresh generator
detector_results = model.predict(
    source=str(VIDEO_PATH),
    task="pose",
    save=False, # We will manually save frames
    conf=0.25,
    iou=0.7,
    imgsz=640,
    stream=True,
    verbose=False, # Suppress verbose output from model.predict
)

violet_frames_count = 0
violet_detected_in_any_frame = False
for frame_idx, r in enumerate(detector_results):
    # r.plot() returns the annotated frame (NumPy array) in BGR format, with keypoints and bounding boxes
    annotated_frame = r.plot()

    # Check for violence in the current frame
    frame_violence_detected = False
    if r.keypoints is not None and len(r.keypoints) > 0:
        for person_kpts in r.keypoints:
            if is_violence_detected(person_kpts, confidence_threshold=0.7):
                frame_violence_detected = True
                violet_detected_in_any_frame = True
                break # If any person shows violence, flag the frame

    if frame_violence_detected:
        violet_frames_count += 1
        # Draw "Violence Detected!" text on the frame
        text = "Violence Detected!"
        font = cv2.FONT_HERSHEY_SIMPLEX
        font_scale = 1.5
        font_thickness = 3
        text_color = (0, 0, 255) # Red in BGR
        text_bg_color = (0, 0, 0) # Black background

        # Get text size for background rectangle
        (text_width, text_height), baseline = cv2.getTextSize(text, font, font_scale, font_thickness)

        # Position for text (top-left corner, with some padding)
        text_x = 20
        text_y = text_height + 20

        # Draw a black rectangle behind the text for better visibility
        cv2.rectangle(annotated_frame, (text_x - 10, text_y - text_height - 10),
                      (text_x + text_width + 10, text_y + 10), text_bg_color, -1)
        cv2.putText(annotated_frame, text, (text_x, text_y), font, font_scale, text_color, font_thickness, cv2.LINE_AA)
        print(f"\033[91mViolence Detected! in Frame {frame_idx + 1}\033[0m") # Red console output

    # Write the (potentially custom-annotated) frame to the output video
    out_video.write(annotated_frame)

out_video.release() # Release the VideoWriter
print(f"Annotated video saved to: {output_video_path}")

# Upload do vídeo anotado (violência) para o bucket S3 yolo-output-tcf4
output_bucket = "yolo-output-tcf4"
if output_video_path.is_file():
    if not all([aws_access_key_id, aws_secret_access_key, aws_default_region]):
        print("S3: credenciais AWS ausentes; vídeo de saída não foi enviado.")
    else:
        s3_client = boto3.client(
            "s3",
            aws_access_key_id=aws_access_key_id,
            aws_secret_access_key=aws_secret_access_key,
            region_name=aws_default_region,
        )
        s3_key = output_video_path.name
        print(f"Enviando '{output_video_path}' para s3://{output_bucket}/{s3_key} ...")
        s3_client.upload_file(str(output_video_path), output_bucket, s3_key)
        print(f"Upload concluído: s3://{output_bucket}/{s3_key}")
else:
    print(f"S3: arquivo de saída não encontrado: {output_video_path}")

if not violet_detected_in_any_frame:
    print("\033[92mNo violence detected based on the current placeholder logic.\033[0m") # Green color
else:
    print(f"\033[91mViolence detected in {violet_frames_count} frames.\033[0m")

print("\033[96mInference with violence detection and video annotation finished.\033[0m") # Cyan color
```
